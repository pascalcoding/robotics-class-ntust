```python
def sign(input):
    return input / abs(input) if input != 0 else 0
# avoid divided by 0

def maxNum(input):
    if abs(input) > 100:
        return 100 * sign(input)
    else:
        return input
```

```python
import ipywidgets.widgets as widgets
# create two sliders with range [-100, 100]
left_slider = widgets.FloatSlider(description='left', min=-100.0, max=100.0, step=1, orientation='vertical')
right_slider = widgets.FloatSlider(description='right', min=-100.0, max=100.0, step=1, orientation='vertical')
# create a horizontal box container to place the sliders next to each other
slider_container = widgets.HBox([left_slider, right_slider])
# display the container in this cell's output
display(slider_container)
```

```python
robot.set_motors(left_slider.value, right_slider.value)
print(left_slider.value, right_slider.value)
time.sleep(5)
robot.stop()
```
